```python
%load_ext autoreload
%autoreload 2

import xp
import time
from pathlib import Path
import scipy.io as sio
from datetime import datetime

from prepare_my_data import GlobalConfig
import run_navigation_trail as run_trail


import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.express as px
import copy

import numpy as np


def compute_metrics(lengths, gt_length):
    """
    Compute statistical metrics for trajectory length estimation.

    Parameters:
        lengths (list or np.ndarray): Estimated trajectory lengths (m)
        gt_length (float): Ground truth length (m)

    Returns:
        dict: Dictionary containing all metrics
    """

    results = {}
    for key_i, lengths_i in lengths.items():

        # Convert to numpy array
        lengths_i = np.array(lengths_i)
        N = len(lengths_i)

        # Compute errors
        errors = lengths_i - gt_length
        abs_errors = np.abs(errors)
        sq_errors = errors ** 2
        rel_errors = errors / gt_length
        abs_rel_errors = np.abs(rel_errors)

        # ----- Basic statistics -----
        mean_length = np.mean(lengths_i)
        std_length = np.std(lengths_i, ddof=1)  # sample standard deviation

        # ----- Error metrics -----
        bias = np.mean(errors)                         # Mean error (bias)
        mae = np.mean(abs_errors)                      # Mean Absolute Error
        mse = np.mean(sq_errors)                       # Mean Squared Error
        rmse = np.sqrt(mse)                            # Root Mean Squared Error

        # ----- Relative error metrics -----
        mean_rel_error = np.mean(rel_errors)
        mape = np.mean(abs_rel_errors) * 100           # Mean Absolute Percentage Error (%)

        # ----- Stability metrics -----
        cv = std_length / mean_length                  # Coefficient of Variation

        # ----- Distribution metrics -----
        max_error_index = np.argmax(abs_errors)
        max_error = abs_errors[max_error_index]

        min_error = np.min(abs_errors)

        p50 = np.percentile(abs_errors, 50)            # Median absolute error
        p90 = np.percentile(abs_errors, 90)
        p95 = np.percentile(abs_errors, 95)

        # ----- Collect results -----
        results_i = {
            "N": N,
            "distance_all": lengths_i,
            "abs_errors": abs_errors,

            "mean_length": mean_length,
            "std_length": std_length,

            "bias": bias,
            "mae": mae,
            "mse": mse,
            "rmse": rmse,

            "mean_relative_error": mean_rel_error,
            "mape_percent": mape,

            "cv": cv,

            "max_error": (max_error_index+1,max_error),
            "min_error": min_error,

            "p50_abs_error": p50,
            "p90_abs_error": p90,
            "p95_abs_error": p95
        }

        results.update({key_i:results_i})

    return results

from typing import Union, List, Dict, Any
def print_metrics(results: Union[Dict[str, Any], Dict[str,Dict[str, Any]]]):
    """
    Pretty-print metrics in a readable format.
    """

    if isinstance(next(iter(results.values())), dict):
        pass
    else:
        results = {"":results}
        

    fig = go.Figure()
    import pandas as pd
    report = []
    for key_i, results_i in results.items():
        data = np.array(results_i["abs_errors"])
        data_sorted = np.sort(data)
        cdf = np.arange(1, len(data_sorted) + 1) / len(data_sorted)

        fig.add_trace(
            go.Scatter(
                x=data_sorted,
                y=cdf,
                mode='lines',
                # line_shape='hv',
                name=key_i
            )
        )

        row = {
            "Device": key_i,
            "N": results_i.get("N"),
            "Mean Length (m)": results_i.get("mean_length"),
            "Std Length (m)": results_i.get("std_length"),
            "Bias (m)": results_i.get("bias"),
            "MAE (m)": results_i.get("mae"),
            "RMSE (m)": results_i.get("rmse"),
            "Mean Rel Error": results_i.get("mean_relative_error"),
            "MAPE (%)": results_i.get("mape_percent"),
            "Coefficient of Variation (CV)": results_i.get("cv"),
            "Max Error (m)": [int(results_i.get("max_error")[0]),float(results_i.get("max_error")[1])],
            "Min Error (m)": results_i.get("min_error"),
            "P50 Error (m)": results_i.get("p50_abs_error"),
            "P90 Error (m)": results_i.get("p90_abs_error"),
            "P95 Error (m)": results_i.get("p95_abs_error"),
        }
        report.append(row)
    
    df = pd.DataFrame(report)
    # df.set_index("Device", inplace=True)
    
    print("\n" + "="*80)
    print("Metrics Comparison:")
    print(df.round(4).to_markdown())
    print("="*80)

    fig.update_layout(
        title="MAE cdf",
        # width=500,
        showlegend=True
    )

    fig.show()

```

```python
dists = ["100cm_3midstop"] #"10cm" ,"1meter", "100cm_3midstop","40cm"
SSMs = ["HYBVIO", "SOLA", "OPENSHOE", "MATERN"] #

all_model_metrics = {}

results_dir = Path(f"results_{datetime.now():%Y%m%d_%H%M%S}")
results_dir.mkdir(parents=True, exist_ok=True)
for dist in dists:
    metrics_by_model = {}
    for SSM_name in SSMs:

        folders = [Path("./data/Xsens_A_" + dist)]
        # folders = [Path("./phyphox_A14A_" + dist),
        #         Path("./phyphox_A14SF_" + dist),
        #         Path("./phyphox_GalaxyS8_" + dist),
        #         Path("./phyphox_Realme_" + dist),
        #         Path("./Xsens_A_" + dist),
        #         Path("./Xsens_B_" + dist)]


        print(f"True distance: {dist} || model: {SSM_name}")

        estimated_lengths = {}
        for folder_i in folders:
            print(f"Processing folder {folder_i}")
            # fixed distance lines
            cfg_glb = GlobalConfig(nav_mode=True, SSM_name=SSM_name, 
                                    metric_type="NLL", #"FRECHET", #
                                    still_time=150,
                                    # ZVDtype="SHOE", zupt_threshold = 150,
                                    ZVDtype="STRICT", zupt_threshold=0.005,
                                    min_zupt_duration_s=1.5,
                                    use_reference_zupt=False,                       
                                    fs=120,      
                                    plot=False                      
                                        )

            cfg_glb.imu_folder = folder_i
                                
            cfg_glb.ref_grid_path = xp.array([[0,0,0],[1,0,0]]).T
            if "1meter" in dist or "100cm" in dist: 
                cfg_glb.grid_scale = 1
            elif "40cm" in dist:
                cfg_glb.grid_scale = 0.4
            elif "25cm" in dist:
                cfg_glb.grid_scale = 0.25
                cfg_glb.still_time = 50
                cfg_glb.min_zupt_duration_s = 0.5
            elif "10cm" in dist:
                cfg_glb.grid_scale = 0.1
                cfg_glb.zupt_threshold=0.003
                cfg_glb.min_zupt_duration_s = 0.5
            elif "60cm" in dist:
                cfg_glb.grid_scale = 0.6
            else:
                cfg_glb.grid_scale = 0

            file_count = len(list(cfg_glb.imu_folder.glob("*.json")))
            print(f"found: {file_count} files in folder {cfg_glb.imu_folder}")
            points_number_for_align = 0


            # Run for specific trails
            trails = [f"{k+1:03d}" for k in range(file_count)] #[f"{k+41:03d}" for k in range(file_count)] #
            cfg_glb.plot = False
            fig_p = make_subplots(
                                rows=1, cols=2,
                                subplot_titles=("2D View", "3D View"),
                                specs=[[{"type": "xy"}, {"type": "scene"}]],
                                column_widths=[0.4, 0.6]
                            )
            fig_p_zupt = go.Figure()
            dist2D_all = []
            dist3D_all = []
            for i, trail in enumerate(trails):
                trail_start = time.time()
                print(f"\n{'='*60}")
                print(f"Processing Trail {trail}")
                print(f"{'='*60}")

                cfg_glb.update({"trail_id":trail})
                            
                try:
                    #  ------  nav mode  ------
                    run_trail.trail_time_start = time.time()

                    # When use_reference_zupt=True, my_out is for using external zupt, my_out_more is for using traditional ZVD
                    # When use_reference_zupt=False, my_out is for using traditional ZVD, my_out_more is None
                    my_out, my_out_more = run_trail.run_navigation_trail(cfg_glb, nav_flag = True)
                    
                    if my_out_more.my_sins is not None:
                        my_out.my_sins.plot_zupt_detection(my_out.my_sins.zupt_ref, u = {"Acc norm":80*xp.norm(my_out.my_sins.u_all[0:3, :], axis=0)})
                        fig1 = my_out.my_sins.plot_quaternion(my_out.my_results, title="Quaternion estimated using reference zupt")
                        fig2 = my_out_more.my_sins.plot_quaternion(my_out_more.my_results,
                                                                    title="Quaternion estimated using traditional ZVD")
                        pts_static_refzupt, fig1 = my_out.my_sins.plot_trajectory_zupt_only(my_out.my_results, title="Static points estimated using reference zupt", points_number_for_align=points_number_for_align)
                        pts_static_zvd, fig2 = my_out_more.my_sins.plot_trajectory_zupt_only(my_out_more.my_results, title="Static points estimated using traditional ZVD", points_number_for_align=points_number_for_align)
                        my_out.my_sins.plot_trajectory(my_out.my_results, title="Trajectory estimated using reference zupt", points_number_for_align=points_number_for_align)
                        my_out_more.my_sins.plot_trajectory(my_out_more.my_results, title="Trajectory estimated using traditional ZVD", points_number_for_align=points_number_for_align)
                        
                        
                    else:
                        # my_out.my_sins.plot_zupt_detection(title=f"{trail}")#u = {"Acc norm":xp.norm(my_out.my_sins.u_all[0:3, :], axis=0)}, 
                        # my_out.my_sins.plot_orientation(my_out.my_results, title=f"{trail}")
                        _, fig1 = my_out.my_sins.plot_trajectory_zupt_only(my_out.my_results, points_number_for_align=0, showfig=False, title=f"{trail}")
                        fig2 = my_out.my_sins.plot_trajectory(my_out.my_results, points_number_for_align=0, showfig=False, title=f"{trail}")
                        
                        dist3D = xp.norm(my_out.my_results["POS"][:,-1]-my_out.my_results["POS"][:,0])
                        dist2D = xp.norm(my_out.my_results["POS"][0:1,-1]-my_out.my_results["POS"][0:1,0])
                        print(f"horizontal dist: {dist2D}, 3D distance: {dist3D}")

                        dist2D_all.append(dist2D)
                        dist3D_all.append(dist3D)
                    

                    trail_lasting = time.time() - trail_start


                    if i==0:
                        fig_p_zupt = copy.deepcopy(fig1)
                        name_mapping = {
                            "ZUPT Interval Means": "trail_"+f"{trail}",
                        }
                        for trace in fig_p_zupt.data:
                            if trace.name in name_mapping:
                                trace.name = name_mapping[trace.name]


                        fig_p = copy.deepcopy(fig2)
                        name_mapping = {
                            "Estimation": f"trail_"+f"{trail}"+"(filter)",
                            "Smoothed estimation": f"trail_"+f"{trail} (smoother)",
                        }
                        for trace in fig_p.data:
                            if trace.name in name_mapping:
                                trace.name = name_mapping[trace.name]

                        continue
                    

                    for t in [copy.deepcopy(t) for t in fig1.data if t.name == "ZUPT Interval Means"]:
                        t.name = f"trail_{trail}"
                        color_trail = px.colors.sample_colorscale('Rainbow', [(int(trail)-int(trails[0]))/file_count])
                        if getattr(t, "line", None): t.line.color = color_trail[0]

                        if getattr(t, "marker", None): t.marker.color = color_trail[0]
                        fig_p_zupt.add_trace(t)

                    for t in [copy.deepcopy(t) for t in fig2.data if t.legendgroup=="Smoothed estimation"]:
                        color_trail = px.colors.sample_colorscale('Rainbow', [(int(trail)-int(trails[0]))/file_count])
                        t.name = f"trail_{trail}"
                        t.legendgroup = f"trail_{trail}"
                        # t.line.color = t.marker.color = color_trail[0]
                        t.mode = 'lines'
                        t.line.color = color_trail[0]
                        if hasattr(t, 'marker'):t.marker = None 
                        fig_p.add_trace(t)         


                    estimated_lengths.update({folder_i.stem: dist3D_all})
                    print(f"\nTrail {trail} completed successfully!")
                    
                except Exception as e:
                    print(f"\nError processing trail {trail}:")
                    print(f"{type(e).__name__}: {e}")
                    trail_lasting = 0
                    import traceback
                    traceback.print_exc()
                
                print(f"Total time: {trail_lasting/60:.2f} minutes")

            fig_p_zupt.update_layout(title_text=f"All static points connection, {cfg_glb.imu_folder.stem}")
            fig_p_zupt.show()
            fig_p.update_layout(title_text=f"All trajectories, {cfg_glb.imu_folder.stem}")
            fig_p.show()



        dist_metrics = compute_metrics(estimated_lengths, gt_length=cfg_glb.grid_scale)
        metrics_by_model.update({SSM_name: next(iter(dist_metrics.values()))})

        var_name = f"{cfg_glb.SSM_name}_{dist}"  
        np.save(results_dir / f"{var_name}.npy", dist_metrics, allow_pickle=True)

    all_model_metrics[dist] = metrics_by_model
```

```python

import winsound
import time
import threading

def melodic_sound(stop_event):
    melody = [(440, 400), (494, 400), (523, 600)]
    while not stop_event.is_set():
        for freq, duration in melody:
            if stop_event.is_set(): return
            winsound.Beep(freq, duration)
            time.sleep(1-duration/1000) 


stop_flag = threading.Event()
thread = threading.Thread(target=melodic_sound, args=(stop_flag,), daemon=True)
thread.start()

print("✅ Main codes running done!")
print("🎵 Playing music")
print("=" * 50)
print("Press ""Enter"" to stop the music")
print("=" * 50)

input()

stop_flag.set()
thread.join(timeout=1)
print("🛑 Music stopped")
```
